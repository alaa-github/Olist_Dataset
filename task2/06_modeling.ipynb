```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

# 1. Load final features
X_train = pd.read_csv('X_train_final.csv')
y_train = pd.read_csv('y_train_final.csv')

# 2. Train a balanced Logistic Regression model
model = LogisticRegression(class_weight='balanced', random_state=42)
model.fit(X_train, y_train.values.ravel())

# 3. Print training performance
train_preds = model.predict(X_train)
print("Baseline Model Performance (Training Set):")
print(classification_report(y_train, train_preds))

# 4. Save the trained model
import joblib
joblib.dump(model, 'logistic_model.joblib')
print("Model trained and saved as logistic_model.joblib")
```
