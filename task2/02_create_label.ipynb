```python
import pandas as pd

# 1. Load the master table
df = pd.read_csv('order_master.csv')

# 2. Convert date columns to datetime objects
df['order_delivered_customer_date'] = pd.to_datetime(df['order_delivered_customer_date'])
df['order_estimated_delivery_date'] = pd.to_datetime(df['order_estimated_delivery_date'])

# 3. Filter only delivered orders (outcome must be known)
df = df.dropna(subset=['order_delivered_customer_date'])

# 4. Create target label: 1 if actual delivery > estimated delivery, else 0
df['is_late'] = (df['order_delivered_customer_date'] > df['order_estimated_delivery_date']).astype(int)

# 5. Save the labeled artifact
df.to_csv('labeled_data.csv', index=False)
print("Target distribution:")
print(df['is_late'].value_counts(normalize=True))
```
