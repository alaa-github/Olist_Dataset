```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Load training data
train = pd.read_csv('train.csv')

# 2. Visualize target distribution
sns.countplot(x='is_late', data=train)
plt.title('Target Distribution (0: On-time, 1: Late)')
plt.savefig('target_distribution.png')
plt.show()

# 3. Analyze relationship between price and delay
sns.boxplot(x='is_late', y='total_price', data=train)
plt.ylim(0, 600)
plt.title('Order Price vs Lateness')
plt.savefig('price_analysis.png')
plt.show()
```
