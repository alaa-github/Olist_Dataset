```python
import pandas as pd

# 1. Load and sort data by purchase time
df = pd.read_csv('labeled_data.csv').sort_values('order_purchase_timestamp')

# 2. Define split points (70% train, 15% val, 15% test)
n = len(df)
train = df.iloc[:int(n*0.7)]
val = df.iloc[int(n*0.7):int(n*0.85)]
test = df.iloc[int(n*0.85):]

# 3. Save split artifacts
train.to_csv('train.csv', index=False)
val.to_csv('val.csv', index=False)
test.to_csv('test.csv', index=False)
print(f"Split complete: Train({len(train)}), Val({len(val)}), Test({len(test)})")
```
