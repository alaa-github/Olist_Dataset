```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
import joblib

def extract_features(file_path):
    data = pd.read_csv(file_path)
    
    # Select basic numerical features
    feature_cols = ['total_price', 'total_freight', 'items_count', 'total_payment', 'max_installments']
    X = data[feature_cols].fillna(0)
    y = data['is_late']
    return X, y

# 1. Process train and validation sets
X_train, y_train = extract_features('train.csv')
X_val, y_val = extract_features('val.csv')

# 2. Scale numerical features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)

# 3. Save processed features and the scaler object
pd.DataFrame(X_train_scaled).to_csv('X_train_final.csv', index=False)
y_train.to_csv('y_train_final.csv', index=False)
joblib.dump(scaler, 'scaler.joblib')
print("Features are engineered and saved.")
```
